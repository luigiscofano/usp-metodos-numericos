## 1. Escalares, vetores e funções
Um array NumPy permite aplicar operações sobre um conjunto de amostras. A expressão `x**2` eleva cada entrada ao quadrado; `x @ x` calcula um produto interno. A indexação começa em zero.

```python
x = np.linspace(-2, 2, 9)
def f(t):
    return t**2 - 2
print("x =", x)
print("f(x) =", f(x))
print("produto interno =", x @ x)
assert x.shape == (9,) and np.isclose(x @ x, 15)
```

## 2. Laço e vetorização
Compare resultados antes de comparar desempenho. Uma versão vetorizada pode usar memória temporária; não é automaticamente a melhor opção em qualquer dimensão.

```python
por_laco = np.array([f(t) for t in x])
vetorizado = f(x)
assert np.array_equal(por_laco, vetorizado)
mask = vetorizado < 0
print("Amostras com f(x)<0:", x[mask])
plt.plot(x, vetorizado, "o-", label="x²−2")
plt.axhline(0, color="black", lw=0.7)
plt.xlabel("x"); plt.ylabel("f(x)"); plt.legend()
mostrar("01_python")
```

## 3. Erro absoluto e relativo
Para referência não nula, o erro relativo é `|aproximação−referência|/|referência|`. Quando a referência é zero, reportar erro absoluto evita divisão sem significado. `isclose` depende de tolerâncias e não é uma igualdade matemática.

```python
reference = np.sqrt(2)
approximation = 1.4142
absolute = abs(approximation-reference)
relative = absolute/abs(reference)
print(f"erro absoluto={absolute:.3e}; erro relativo={relative:.3e}")
print("0.1+0.2 == 0.3?", 0.1+0.2 == 0.3)
assert np.isclose(0.1+0.2,0.3,rtol=0,atol=1e-15)
assert absolute < 2e-5
```
